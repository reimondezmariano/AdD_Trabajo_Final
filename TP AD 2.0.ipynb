# Dataset Elegido:

## Diabetes 130-US Hospitals for Years 1999-2008

El conjunto de datos abarca diez años (1999-2008) de atención clínica en **130 hospitales y redes integradas de prestación de servicios sanitarios de EE. UU**. Cada fila corresponde a los historiales hospitalarios de pacientes diagnosticados con diabetes, a los que se les realizaron análisis de laboratorio, se les **administraron medicamentos y permanecieron ingresados hasta 14 días**. El objetivo es **determinar el reingreso precoz del paciente en los 30 días siguientes al alta**. El problema es importante por las siguientes razones. A pesar de que existen pruebas de alta calidad que demuestran una mejora en los resultados clínicos de los pacientes diabéticos que reciben diversas intervenciones preventivas y terapéuticas, **muchos pacientes no las reciben**. Esto puede atribuirse en parte a un tratamiento arbitrario de la diabetes en el ámbito hospitalario, que no presta la atención necesaria al control glucémico. La falta de una atención adecuada de la diabetes no solo aumenta los costes de gestión para los hospitales (al producirse **reingresos de los pacientes**), sino que también repercute en la morbilidad y la mortalidad de los pacientes, que pueden sufrir complicaciones asociadas a la diabetes.

Información adicional

El dataset representa diez años (1999–2008) de atención clínica en 130 hospitales y redes integradas de salud de Estados Unidos. Incluye más de 50 variables que representan resultados de pacientes y hospitales. La información fue extraída de la base de datos para los encuentros (episodios de atención) que cumplían con los siguientes criterios:

1. Es un encuentro de internación (una admisión hospitalaria).
2. Es un encuentro diabético, es decir, uno durante el cual se registró en el sistema algún tipo de diabetes como diagnóstico.
3. La duración de la internación fue de al menos 1 día y como máximo 14 días.
4. Se realizaron análisis de laboratorio durante el encuentro.
5. Se administraron medicamentos durante el encuentro.

Los datos contienen atributos como número de paciente, raza, género, edad, tipo de admisión, tiempo en el hospital, especialidad médica del médico que admitió al paciente, cantidad de análisis de laboratorio realizados, resultado del test de HbA1c, diagnóstico, cantidad de medicamentos, medicamentos para la diabetes, cantidad de visitas ambulatorias, de internación y de emergencia en el año previo a la hospitalización, etc.

### 2.2 Decisiones de tratamiento

```python
faltantes_pct = (data_set.isnull().sum() / len(data_set) * 100)

# --- weight: se elimina la columna (97% vacía = no aporta información explotable) ---
data_set = data_set.drop(columns=["weight"])

# --- max_glu_serum / A1Cresult: el faltante ES información (no se pidió el análisis).
#     Se imputa con una categoría explícita "No medido" en lugar de perder la variable ---
for c in ["max_glu_serum", "A1Cresult"]:
    data_set[c] = data_set[c].astype("object").fillna("No medido").astype("category")

# --- medical_specialty / payer_code: alta cardinalidad + muchos NaN -> categoría "Faltante" ---
for c in ["medical_specialty", "payer_code"]:
    data_set[c] = data_set[c].astype("object").fillna("Faltante").astype("category")

# --- race: baja proporción de nulos, variable categórica nominal -> imputar con la moda ---
moda_race = data_set["race"].mode()[0]
data_set["race"] = data_set["race"].astype("object").fillna(moda_race).astype("category")

# --- diag_1/2/3: muy pocos nulos, no se pueden imputar con sentido clínico -> se dejan como
#     categoría "Sin diagnostico" para no perder filas ---
for c in ["diag_1", "diag_2", "diag_3"]:
    data_set[c] = data_set[c].astype("object").fillna("Sin_diagnostico").astype("category")

print("Nulos restantes:", data_set.isnull().sum().sum())
print("Nueva forma del dataset:", data_set.shape)
```

**Justificación general:** para variables numéricas hubiéramos usado media/mediana según

por lo que se optó por: (a) eliminar la columna cuando la pérdida de información es casi total
(`weight`), o (b) crear una categoría explícita de "faltante/no medido" cuando el hecho de que
falte el dato es en sí mismo informativo (ej. que no se haya pedido A1C), evitando así el sesgo
de imputar con la moda en variables de alta cardinalidad.

También se elimina `gender == "Unknown/Invalid"` (sólo 3 casos) por ser un valor
inconsistente sin categoría clínica real.

```python
print("Casos con gender inválido:", (data_set["gender"] == "Unknown/Invalid").sum())
data_set = data_set[data_set["gender"] != "Unknown/Invalid"].copy()
data_set["gender"] = data_set["gender"].cat.remove_unused_categories()
data_set.shape
```

### 2.3 Detección de outliers (método IQR)

Se aplica el criterio de rango intercuartílico (IQR) sobre las variables numéricas más
relevantes clínicamente: `num_lab_procedures` (cantidad de análisis de laboratorio),
`number_emergency` (visitas a emergencias en el año previo), `number_inpatient`
(internaciones previas) y `num_medications` (cantidad de medicamentos administrados).

```python
def resumen_outliers_iqr(serie):
    q1, q3 = serie.quantile([0.25, 0.75])
    iqr = q3 - q1
    lim_inf, lim_sup = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    outliers = serie[(serie < lim_inf) | (serie > lim_sup)]
    return lim_inf, lim_sup, len(outliers), len(outliers) / len(serie) * 100

for col in ["num_lab_procedures", "number_emergency", "number_inpatient", "num_medications"]:
    lim_inf, lim_sup, n_out, pct_out = resumen_outliers_iqr(data_set[col])
    print(f"{col:22s} límites=({lim_inf:6.1f}, {lim_sup:6.1f})  outliers={n_out:6d} ({pct_out:5.2f}%)  max={data_set[col].max()}")
```

```python
fig, axes = plt.subplots(1, 4, figsize=(18, 4))
for ax, col in zip(axes, ["num_lab_procedures", "number_emergency", "number_inpatient", "num_medications"]):
    sns.boxplot(y=data_set[col], ax=ax, color="skyblue")
    ax.set_title(col)
plt.tight_layout()
plt.show()
```

#### Método alternativo: desvío estándar (±3σ), como contraste con IQR

Se compara el criterio de IQR con el criterio clásico de distribución normal
(outlier si está a más de 3 desvíos estándar de la media), para ver si ambos métodos coinciden.

```python
def resumen_outliers_std(serie):
    media, std = serie.mean(), serie.std()
    lim_inf, lim_sup = media - 3 * std, media + 3 * std
    outliers = serie[(serie < lim_inf) | (serie > lim_sup)]
    return lim_inf, lim_sup, len(outliers), len(outliers) / len(serie) * 100

comparacion = []
for col in ["num_lab_procedures", "number_emergency", "number_inpatient", "num_medications"]:
    _, _, n_iqr, pct_iqr = resumen_outliers_iqr(data_set[col])
    _, _, n_std, pct_std = resumen_outliers_std(data_set[col])
    comparacion.append([col, n_iqr, round(pct_iqr, 2), n_std, round(pct_std, 2)])

pd.DataFrame(comparacion, columns=["variable", "outliers_IQR", "%_IQR", "outliers_3sigma", "%_3sigma"])
```

**Lectura:** el método de ±3σ marca sistemáticamente **muchos menos outliers** que IQR en estas
variables. Tiene sentido: ±3σ asume una distribución aproximadamente normal, y estas variables
tienen colas muy largas y asimétricas (ya cuantificado con `skew`/`kurtosis` en la sección 3.1),
lo que hace que la media y el desvío estándar estén "inflados" por esa misma cola -subestimando
cuántos valores son realmente atípicos. Para este dataset, **IQR es el criterio más apropiado**
por no asumir normalidad.

**Interpretación clínica de los outliers:**
- `num_lab_procedures`: el IQR marca ~143 casos como outliers (0.14%), pero el máximo observado
  (132) es perfectamente plausible en una internación larga con múltiples estudios — **no se
  eliminan**, son pacientes con cuadros complejos que requieren muchos análisis.
- `number_emergency` (11.19% marcado como outlier) y, en menor medida, `number_inpatient`
  (6.93%): la mayoría de los pacientes tiene 0 visitas previas, por lo que el IQR marca como
  "outlier" a *cualquier valor positivo* en estas variables. Esto es un artefacto de una
  distribución muy asimétrica (cola larga), no un error de carga. Son justamente los pacientes
  de alto riesgo — la población que más interesa para el problema de reingreso, así que
  **tampoco se eliminan**.
- `num_medications` (2.51% marcado como outlier, máximo observado 81): asimetría más moderada
  que las anteriores, pero con la misma lógica — valores altos reflejan cuadros con polimedicación,
  clínicamente relevantes y no errores de carga.
- **Decisión final:** no se eliminan outliers ni se aplica winsorización agresiva, porque en
  este dominio los valores extremos representan pacientes clínicamente críticos y son la señal
  más relevante para predecir reingreso. En su lugar, se aplica **winsorización suave (capping
  al percentil 99)** sólo para evitar que valores extremos aislados distorsionen el escalado
  posterior de los modelos.

```python
for col in ["num_lab_procedures", "number_emergency", "number_inpatient", "num_medications"]:
    p99 = data_set[col].quantile(0.99)
    data_set[col] = np.where(data_set[col] > p99, p99, data_set[col])

data_set[["num_lab_procedures", "number_emergency", "number_inpatient", "num_medications"]].describe()
```

#### Alternativa considerada: transformación logarítmica

Como alternativa a la winsorización, se evalúa transformar `number_emergency` con `log1p`
(usa `log(1+x)` para poder incluir los ceros, que son mayoría en esta variable).

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(data_set["number_emergency"], bins=30, ax=axes[0], color="steelblue")
axes[0].set_title("number_emergency (winsorizada, escala original)")

log_emergency = np.log1p(data_set["number_emergency"])
sns.histplot(log_emergency, bins=30, ax=axes[1], color="salmon")
axes[1].set_title("number_emergency con transformación log1p")
plt.tight_layout()
plt.show()
```

**Lectura:** la transformación logarítmica comprime la cola y "normaliza" más la forma de la
distribución, lo cual puede ayudar a modelos sensibles a la escala (ej. regresión lineal/logística
sin regularización). Sin embargo, **se descarta para este trabajo** por dos motivos: (1) más del
90% de los valores son 0, y `log1p` los sigue dejando en 0, sin resolver el problema real (la
masa puntual en cero domina igual la distribución); y (2) para los modelos basados en árboles
(Random Forest, Gradient Boosting) -candidatos naturales para este problema- la escala de las
variables no afecta el desempeño, así que la ganancia de transformar no compensa la pérdida de
interpretabilidad directa de la variable (mejor hablar de "2 internaciones previas" que de su
logaritmo). Se mantiene la decisión de winsorización suave ya aplicada.

## 2.4 Recodificación de variables

Con el dataset ya limpio de faltantes y outliers (secciones anteriores), se recodifican
las variables restantes teniendo en cuenta su naturaleza real (nominal vs. ordinal), para
que el EDA de la sección siguiente ya trabaje sobre variables correctamente tipadas.

### `patient_nbr` -> variable derivada ordinal

`patient_nbr` en si mismo es un identificador (sin orden real entre sus valores), por lo
que no se recodifica directamente. En cambio, se deriva una variable con sentido clinico
real y orden natural: la **cantidad de encuentros PREVIOS que tuvo cada paciente** antes
de la fila actual, agrupada en bins ordinales.

Es importante contar solo encuentros previos (no el total de encuentros del paciente en
todo el dataset): usar el total incluiria informacion de encuentros futuros respecto a la
fila actual, lo cual seria una fuga de informacion hacia el futuro (data leakage) si la
variable se usa como feature para predecir el reingreso. Para ordenar los encuentros de
cada paciente en el tiempo se usa `encounter_id`, que dentro de un mismo paciente es
estrictamente creciente y funciona como proxy cronologico (el dataset no trae una fecha
explicita).

```python
# 1) Ordenar por paciente y encounter_id (proxy de orden cronologico dentro de cada paciente)
#    y contar, para cada fila, cuantos encuentros PREVIOS tuvo ese paciente (sin incluir el actual
#    ni los futuros), para evitar fuga de informacion hacia el futuro.
data_set = data_set.sort_values(["patient_nbr", "encounter_id"]).reset_index(drop=True)
data_set["num_encuentros_previos"] = data_set.groupby("patient_nbr").cumcount()

# 2) Agrupar en bins ordinales (0, 1, 2, 3, 4, 5 o mas) segun encuentros previos
def bucketizar(x):
    if x == 0: return "0"
    elif x == 1: return "1"
    elif x == 2: return "2"
    elif x == 3: return "3"
    elif x == 4: return "4"
    else: return "5 o mas"

orden = ["0", "1", "2", "3", "4", "5 o mas"]
data_set["patient_nbr_ordinal"] = pd.Categorical(
    data_set["num_encuentros_previos"].apply(bucketizar),
    categories=orden,
    ordered=True
)

print("dtype:", data_set["patient_nbr_ordinal"].dtype)
print("ordered:", data_set["patient_nbr_ordinal"].cat.ordered)
data_set["patient_nbr_ordinal"].value_counts(normalize=True).reindex(orden).mul(100).round(2)
```

### 3.2 Análisis univariado — variables categóricas (cardinalidad y entropía de Shannon)

La entropía de Shannon mide qué tan "repartidas" están las observaciones entre categorías:
cerca de 0 = una categoría domina casi todo; cerca del máximo (log2 del nº de categorías) =
distribución perfectamente equilibrada.

```python
def entropia_shannon(serie):
    counts = serie.value_counts(dropna=True)
    probs = counts / counts.sum()
    h = stats.entropy(probs, base=2)
    h_max = np.log2(len(counts))
    return counts.shape[0], h, h_max, (h / h_max if h_max > 0 else np.nan)

cols_categoricas_interes = ["race", "gender", "age", "admission_type_id",
                              "discharge_disposition_id", "admission_source_id",
                              "medical_specialty", "payer_code", "diag_1", "diag_2", "diag_3",
                              "max_glu_serum", "A1Cresult"]

filas = []
for c in cols_categoricas_interes:
    n_cat, h, h_max, h_norm = entropia_shannon(data_set[c])
    filas.append([c, n_cat, round(h, 2), round(h_max, 2), round(h_norm, 2)])

pd.DataFrame(filas, columns=["variable", "n_categorias", "entropia", "entropia_max", "entropia_normalizada"])
```

**Lectura:** `gender` y `age` tienen entropía normalizada alta (categorías balanceadas).
`race` y `admission_source_id` están dominadas por 1-2 categorías (`Caucasian` y el ingreso por
guardia, respectivamente), lo que se refleja en una entropía normalizada más baja.
`medical_specialty`, con 70+ categorías, tiene alta cardinalidad pero baja entropía relativa:
pocas especialidades concentran la mayoría de los casos (ej. `InternalMedicine`), algo a tener
en cuenta al elegir la estrategia de encoding (sección 4).

#### Versión en gráfico de dona (donut) de la distribución de `race`

Dos variantes: la primera muestra **todas** las categorías tal cual aparecen en el dataset; la segunda agrupa las categorías minoritarias (< 10% cada una) en "Otros", igual que se suele hacer para mejorar la legibilidad cuando hay muchas clases chicas.

```python
def plot_donut(data, column, group_others=False, threshold=0.10,
               title=None, palette="pastel", ax=None, radius=0.85,
               min_gap=20, radio_texto_mult=1.35, fontsize=8.5, title_pad=30):
    """
    Gráfico de dona con etiquetas externas (líneas guía) y total en el centro.
    Cuando varias porciones quedan angularmente muy juntas (categorías chicas
    pegadas entre sí), sus etiquetas se separan automáticamente en ángulo y en
    distancia para no superponerse ni entre sí ni con el título del gráfico.
    """
    counts = data[column].value_counts()

    if group_others:
        pct = counts / counts.sum()
        major = counts[pct >= threshold].copy()
        minor = counts[pct < threshold]
        if len(minor) > 0:
            major["Otros"] = minor.sum()
        counts = major.sort_values(ascending=False)
    else:
        counts = counts.sort_values(ascending=False)

    total = counts.sum()
    colores = sns.color_palette(palette, len(counts))

    if ax is None:
        fig, ax = plt.subplots(figsize=(6, 5.5))

    wedges, _ = ax.pie(
        counts.values,
        startangle=90,
        colors=colores,
        radius=radius,
        wedgeprops=dict(width=0.4 * radius, edgecolor="white", linewidth=1.5),
    )

    angulos_true = np.array([(w.theta2 + w.theta1) / 2 for w in wedges])

    # Relajación 1D: separar los ángulos de etiqueta que quedan a menos de min_gap
    # grados de distancia entre sí (categorías chicas y pegadas no se pisan)
    angulos_disp = angulos_true.copy()
    for _ in range(500):
        cambiado = False
        for i in range(1, len(angulos_disp)):
            gap = angulos_disp[i] - angulos_disp[i - 1]
            if gap < min_gap:
                falta = (min_gap - gap) / 2
                angulos_disp[i] += falta
                angulos_disp[i - 1] -= falta
                cambiado = True
        if not cambiado:
            break

    radio_texto = radius * radio_texto_mult
    for i, wedge in enumerate(wedges):
        ang_true, ang_disp = angulos_true[i], angulos_disp[i]
        x_true, y_true = np.cos(np.radians(ang_true)), np.sin(np.radians(ang_true))
        x_disp, y_disp = np.cos(np.radians(ang_disp)), np.sin(np.radians(ang_disp))

        alineacion = "right" if x_disp < 0 else "left"
        valor = counts.values[i]
        pct = valor / total * 100
        ax.annotate(
            f"{valor:,} ({pct:.1f}%)",
            xy=(x_true * radius * 0.88, y_true * radius * 0.88),
            xytext=(x_disp * radio_texto, y_disp * radio_texto),
            horizontalalignment=alineacion,
            fontsize=fontsize,
            arrowprops=dict(arrowstyle="-", color="gray", lw=0.8,
                             connectionstyle="arc3,rad=0.08"),
        )

    ax.text(0, 0, f"{total:,}\npacientes", ha="center", va="center",
             fontsize=12, fontweight="bold", color="#333333")

    ax.legend(wedges, counts.index, title=column.capitalize(),
               loc="center left", bbox_to_anchor=(1.05, 0.5), frameon=False, fontsize=9)
    # pad más grande: deja aire entre las etiquetas de arriba y el título
    ax.set_title(title if title else f'Distribución de "{column}"',
                  fontsize=13, fontweight="bold", pad=title_pad)
    ax.set_xlim(-2.3, 2.3)
    ax.set_ylim(-2.1, 2.1)
    ax.axis("equal")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 6))
# top más bajo (0.76) para dejarle más aire al título de cada gráfico
fig.subplots_adjust(wspace=0.7, top=0.76, bottom=0.05)

plot_donut(data_set, "race", group_others=False,
           title="Todas las categorías", ax=axes[0])

plot_donut(data_set, "race", group_others=True, threshold=0.10,
           title='Agrupado (< 10% -> "Otros")', ax=axes[1])

# suptitle más arriba (y=1.02) para no pisar los títulos individuales
fig.suptitle('Distribución de "race"', fontsize=15, fontweight="bold", y=1.02)
plt.show()
```

**Balance del target:** se observa a simple vista que `readmitted` está desbalanceado —
la mayoría de los pacientes no vuelve a ser internado (`NO`), y sólo una fracción reingresa
dentro de los 30 días (`<30`). Esto se cuantifica en la sección 6.

#### Test Chi-cuadrado de independencia y V de Cramér

Se mide la asociación entre variables categóricas y el target `readmitted`.

```python
def chi2_cramers_v(col, target="readmitted"):
    tabla = pd.crosstab(data_set[col], data_set[target])
    chi2, p, dof, _ = chi2_contingency(tabla)
    n = tabla.sum().sum()
    k = min(tabla.shape) - 1
    v = np.sqrt(chi2 / (n * k)) if k > 0 else np.nan
    return chi2, p, v

vars_categoricas_test = ["age", "gender", "race", "admission_type_id", "discharge_disposition_id",
                           "admission_source_id", "payer_code", "diag_1", "diag_2", "diag_3",
                           "A1Cresult", "max_glu_serum", "insulin", "diabetesMed", "change"]

resultados = []
for c in vars_categoricas_test:
    chi2, p, v = chi2_cramers_v(c)
    resultados.append([c, round(chi2, 1), p, round(v, 3)])

pd.DataFrame(resultados, columns=["variable", "chi2", "p_valor", "V_de_Cramer"]).sort_values("V_de_Cramer", ascending=False)
```

Todas las variables analizadas resultan **estadísticamente asociadas** al target
(p-valor << 0.05, dado el tamaño muestral tan grande), pero el p-valor por sí solo no indica
*cuán fuerte* es la asociación. La V de Cramér sí lo hace: `discharge_disposition_id` se destaca
claramente por encima del resto (V=0.133). Le sigue `admission_source_id` (V=0.075) y, más atrás,
un grupo de variables con asociación moderada-baja y sin diferencias grandes entre sí:
`diabetesMed`, `diag_1`, `payer_code`, `insulin`, `change`, `admission_type_id` y `age`
(V entre 0.036 y 0.062) — notablemente, `diag_1` no se destaca frente a estas otras una vez que
ya está agrupado por capítulo ICD-9 (11 categorías, sección 2.4), y `diag_2`/`diag_3` quedan en
ese mismo rango bajo. En términos absolutos todas son moderadas-bajas, típico en datos clínicos
con múltiples factores en juego, mientras que `race`, `gender`, `A1Cresult` y `max_glu_serum`
son las que menos discriminan el reingreso (V ≈ 0.016-0.020).

#### Medidas de asociación complementarias (Phi, Punto Biserial, Kendall Tau-b, Eta²)

El test Chi² y la V de Cramér cubren pares **categórica-categórica**. Para caracterizar mejor
la relación entre las variables y el target -y entre sí- conviene usar la medida adecuada según
el tipo de cada variable involucrada. Se adelanta aquí, únicamente a fines exploratorios, la
binarización de `readmitted` (`readmitted_bin_tmp`) que se formaliza en la sección 4.

```python
data_set["readmitted_bin_tmp"] = (data_set["readmitted"] == "<30").astype(int)
```

**1) Coeficiente Phi (binaria - binaria)**

```python
def phi(x, y):
    tabla = pd.crosstab(x, y)
    chi2 = chi2_contingency(tabla)[0]
    n = tabla.sum().sum()
    return np.sqrt(chi2 / n)

vars_binarias = ["gender", "diabetesMed", "change", "readmitted_bin_tmp"]

phi_mat = pd.DataFrame(index=vars_binarias, columns=vars_binarias, dtype=float)
for c1 in vars_binarias:
    for c2 in vars_binarias:
        if c1 == c2:
            phi_mat.loc[c1, c2] = 1.0
        else:
            phi_mat.loc[c1, c2] = phi(data_set[c1], data_set[c2])

mask_upper = np.triu(np.ones_like(phi_mat, dtype=bool))
fig, ax = plt.subplots(figsize=(5, 4))
sns.heatmap(phi_mat.astype(float), annot=True, fmt=".2f", cmap="YlOrRd",
            vmin=0, vmax=1, linewidths=0.5, mask=mask_upper, ax=ax)
ax.set_title("Phi - Asociación binaria-binaria")
ax.tick_params(axis="x", rotation=30)
plt.tight_layout()
plt.show()
```

**Lectura:** `change` y `diabetesMed` muestran la asociación más alta del grupo (esperable:
si no hay medicación para diabetes, tampoco puede haber cambio de medicación). Contra el target
binarizado, ninguna variable binaria por sí sola tiene una Phi alta -consistente con lo visto
en la V de Cramér: el reingreso temprano es multifactorial y no depende fuertemente de una sola
variable binaria.

simultáneos tienen mayor riesgo. Todos los coeficientes son individualmente bajos en magnitud
(< 0.2), lo cual es coherente con el fenómeno multifactorial que se viene observando, pero las

**3) Kendall Tau-b (numérica - ordinal)**

`age` es una variable categórica **ordinal** (los rangos etarios tienen un orden natural), por
lo que Kendall Tau-b es más apropiado que Cramér's V para medir su asociación con las variables
numéricas.

```python
age_orden = sorted(data_set["age"].unique(), key=lambda s: int(s.strip("[)").split("-")[0]))
age_codes = data_set["age"].map({cat: i for i, cat in enumerate(age_orden)})

resultados_tau = []
for col in cols_numericas:
    tau, p = kendalltau(age_codes, data_set[col], variant="b")
    resultados_tau.append([col, round(tau, 3), p])

tau_df = pd.DataFrame(resultados_tau, columns=["variable", "kendall_tau_b", "p_valor"])
tau_df.sort_values("kendall_tau_b", key=abs, ascending=False)
```

**Lectura:** `number_diagnoses` y `time_in_hospital` crecen con la edad (tau positivo):
pacientes mayores acumulan más diagnósticos simultáneos y tienden a internaciones algo más

visitas de urgencia y los procedimientos quirúrgicos/invasivos se concentran algo más en
pacientes jóvenes. El resto de las variables (incluida `number_inpatient`) tiene una asociación
prácticamente nula con la edad. Las magnitudes son en todos los casos bajas, por lo que `age`
aporta información relevante pero no domina la explicación de estas variables.

**4) Eta cuadrado (numérica - categórica)**

Complementa el test Chi²/Cramér's V: mide qué proporción de la varianza de cada variable
numérica es explicada por las 3 clases originales de `readmitted` (`NO`, `>30`, `<30`).

```python
def eta_squared(continua, categorica):
    y, grp = continua, categorica
    grupos = [g.values for _, g in y.groupby(grp, observed=True)]
    F, p = f_oneway(*grupos)
    media_total = y.mean()
    ss_total = ((y - media_total) ** 2).sum()
    ss_between = sum(len(g) * (g.mean() - media_total) ** 2 for g in grupos)
    return ss_between / ss_total, p, F

resultados_eta = []
for col in cols_numericas:
    eta2, p, F = eta_squared(data_set[col], data_set["readmitted"])
    resultados_eta.append([col, round(eta2, 4), p])

eta_df = pd.DataFrame(resultados_eta, columns=["variable", "eta_cuadrado", "p_valor"])
eta_df = eta_df.sort_values("eta_cuadrado", ascending=False)

fig, ax = plt.subplots(figsize=(8, 4.5))
sns.barplot(data=eta_df, y="variable", x="eta_cuadrado", color="mediumpurple", ax=ax)
ax.set_title("Eta cuadrado - variables numéricas vs. readmitted (3 clases)")
ax.set_xlabel("Eta² (proporción de varianza explicada)")
plt.tight_layout()
plt.show()

eta_df
```

**Lectura:** consistente con Punto Biserial, `number_inpatient` concentra claramente la mayor
proporción de varianza explicada por `readmitted` (Eta² ≈ 0.055, en el límite entre "efecto
chico" y "mediano" según Cohen), seguido de cerca por `number_diagnoses` (≈ 0.013) y
`number_emergency` (≈ 0.011). El resto de las variables numéricas quedan por debajo de 0.01
("efecto chico"). Esto refuerza la conclusión general de la sección: **ninguna variable aislada
explica el reingreso temprano**, y el problema requiere un modelo multivariado.

### 3.4 Ingeniería de variables: familias farmacológicas de medicamentos

Las 23 columnas de medicamentos (una por droga) son difíciles de analizar por separado:
la mayoría tiene muy poca variabilidad, y comparar 23 gráficos droga por droga no aporta
una lectura clara. En lugar de eso, se agrupan las drogas por **familia farmacológica**
(clasificación médica estándar, no específica de este dataset) y se colapsa cada familia
en una sola variable con la misma escala original: `No`, `Steady`, `Up`, `Down`.

| Familia | Columnas que la integran |
|---|---|
| Biguanidas | `metformin` |
| Sulfonilureas | `chlorpropamide`, `glimepiride`, `acetohexamide`, `glipizide`, `glyburide`, `tolbutamide`, `tolazamide` |
| Meglitinidas | `repaglinide`, `nateglinide` |
| Glitazonas (tiazolidinedionas) | `pioglitazone`, `rosiglitazone`, `troglitazone` |
| Inhibidores de alfa-glucosidasa | `acarbose`, `miglitol` |
| Insulina | `insulin` |

Las combinaciones de dos drogas (`glyburide-metformin`, `glipizide-metformin`,
`glimepiride-pioglitazone`, `metformin-rosiglitazone`, `metformin-pioglitazone`) aportan
a **las dos familias** que combinan. `examide` y `citoglipton` se excluyen: no fueron
prescriptas a ningún paciente en este dataset (varianza cero).

Cuando una familia tiene más de una droga informada para el mismo paciente, se toma el
estado de mayor "intensidad" con la prioridad `Up > Down > Steady > No`.

```python
# Mapeo droga -> familia(s) farmacológica(s)
mapa_familias = {
    "metformin": ["Biguanidas"],
    "repaglinide": ["Meglitinidas"],
    "nateglinide": ["Meglitinidas"],
    "chlorpropamide": ["Sulfonilureas"],
    "glimepiride": ["Sulfonilureas"],
    "acetohexamide": ["Sulfonilureas"],
    "glipizide": ["Sulfonilureas"],
    "glyburide": ["Sulfonilureas"],
    "tolbutamide": ["Sulfonilureas"],
    "pioglitazone": ["Glitazonas"],
    "rosiglitazone": ["Glitazonas"],
    "acarbose": ["Alfa_glucosidasa"],
    "miglitol": ["Alfa_glucosidasa"],
    "troglitazone": ["Glitazonas"],
    "tolazamide": ["Sulfonilureas"],
    "examide": [],           # varianza cero -> se excluye
    "citoglipton": [],       # varianza cero -> se excluye
    "insulin": ["Insulina"],
    "glyburide-metformin": ["Sulfonilureas", "Biguanidas"],
    "glipizide-metformin": ["Sulfonilureas", "Biguanidas"],
    "glimepiride-pioglitazone": ["Sulfonilureas", "Glitazonas"],
    "metformin-rosiglitazone": ["Biguanidas", "Glitazonas"],
    "metformin-pioglitazone": ["Biguanidas", "Glitazonas"],
}

familias = ["Biguanidas", "Sulfonilureas", "Meglitinidas", "Glitazonas", "Alfa_glucosidasa", "Insulina"]
prioridad = {"Up": 3, "Down": 2, "Steady": 1, "No": 0}

for familia in familias:
    cols_familia = [droga for droga, fams in mapa_familias.items() if familia in fams]
    nombre_col = f"familia_{familia.lower()}"
    data_set[nombre_col] = data_set[cols_familia].apply(
        lambda fila: max(fila, key=lambda v: prioridad.get(v, 0)), axis=1
    )

cols_familias_nuevas = [f"familia_{f.lower()}" for f in familias]
data_set[cols_familias_nuevas].head()
```

## 4. Problema de Machine Learning propuesto

## 5. Preparación para modelos (Data Wrangling & Prep)

### 5.1 Split entrenamiento / prueba (estratificado)

Como el target está desbalanceado (~11% de reingresos tempranos), se usa
`train_test_split(..., stratify=y)` para que ambos conjuntos mantengan la misma proporción de
clases.

```python
id_cols = ["encounter_id", "patient_nbr"]
target_col = "readmitted_temprano"

# Las 23 columnas de droga individual quedan reemplazadas por las 6 variables de
# familia farmacológica (sección 3.4) -- se excluyen de X para no arrastrarlas sin usar.
cols_droga_individual = list(mapa_familias.keys())

drop_cols = id_cols + ["readmitted", target_col] + cols_droga_individual

X = data_set.drop(columns=drop_cols)
y = data_set[target_col]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=RANDOM_STATE, stratify=y
)

print("Train:", X_train.shape, " Test:", X_test.shape)
print("Proporción de clase positiva - train:", y_train.mean().round(3), " test:", y_test.mean().round(3))
```

### 5.2 Codificación (encoding) de variables categóricas

Se eligió la estrategia según la **cardinalidad y el tipo de orden** de cada variable:

| Estrategia | Variables | Motivo |
|---|---|---|
| **One-hot encoding** | `gender`, `A1Cresult`, `max_glu_serum`, `diabetesMed`, `change`, las 6 variables de familia farmacológica (`familia_biguanidas`, etc.), `race`, `payer_code`, `admission_type_id` (baja cardinalidad, sin orden) | Pocas categorías → no explota la dimensionalidad y evita imponer un orden inexistente. |
| **Ordinal encoding** | `age`, `patient_nbr_ordinal` (con orden natural) | Sus categorías tienen un orden lógico creciente; tratarlas como ordinales preserva esa información sin crear columnas dummy. |
| **Target encoding** (media del target por categoría, calculada sólo con train) | `diag_1`, `diag_2`, `diag_3` (11 categorías cada una, tras la agrupación por capítulo ICD-9 de la sección 2.4), `medical_specialty` (72), `discharge_disposition_id` (26), `admission_source_id` (17) | Para `medical_specialty` en particular, one-hot generaría demasiadas columnas dispersas; para el resto, aunque la cardinalidad ya no es tan alta como los códigos ICD-9 originales (`diag_1/2/3` llegaban a 700+ categorías antes de agruparse), el target encoding sigue siendo preferible porque condensa cada categoría directamente en su tasa histórica de reingreso, capturando señal predictiva de forma más directa que el one-hot. Se calcula **únicamente con el set de entrenamiento** para evitar data leakage. |

```python
# --- Ordinal encoding de age (orden natural de los intervalos) ---
orden_age = sorted(X_train["age"].unique(), key=lambda s: int(s.strip("[)>").split("-")[0]))
ordinal_enc = OrdinalEncoder(categories=[orden_age])

X_train["age_ord"] = ordinal_enc.fit_transform(X_train[["age"]])
X_test["age_ord"] = ordinal_enc.transform(X_test[["age"]])

print("Orden de age:", orden_age)

# --- Ordinal encoding de patient_nbr_ordinal (orden natural: 0, 1, 2, 3, 4, 5 o mas -- encuentros PREVIOS a cada fila) ---
orden_patient_nbr = ["0", "1", "2", "3", "4", "5 o mas"]
ordinal_enc_pnbr = OrdinalEncoder(categories=[orden_patient_nbr])

X_train["patient_nbr_ordinal_ord"] = ordinal_enc_pnbr.fit_transform(X_train[["patient_nbr_ordinal"]])
X_test["patient_nbr_ordinal_ord"] = ordinal_enc_pnbr.transform(X_test[["patient_nbr_ordinal"]])

print("Orden de patient_nbr_ordinal:", orden_patient_nbr)
```

```python
# --- Target encoding (calculado SOLO con train, para evitar leakage) para variables de alta cardinalidad ---
cols_alta_cardinalidad = ["diag_1", "diag_2", "diag_3", "medical_specialty",
                            "discharge_disposition_id", "admission_source_id"]
media_global = y_train.mean()

for c in cols_alta_cardinalidad:
    medias_categoria = y_train.groupby(X_train[c], observed=True).mean()
    X_train[c + "_te"] = X_train[c].map(medias_categoria)
    X_test[c + "_te"] = X_test[c].map(medias_categoria).fillna(media_global)  # categorías no vistas -> media global

X_train[[c + "_te" for c in cols_alta_cardinalidad]].describe()
```

#### Alternativas comparadas para las variables de alta cardinalidad: Frequency y Binary Encoding

Antes de continuar con target encoding, se comparan dos alternativas típicas para variables de
alta cardinalidad. `medical_specialty` mantiene sus 72 categorías originales. `diag_1`, `diag_2`
y `diag_3`, en cambio, ya fueron agrupadas por capítulo ICD-9 en la sección 2.4 y hoy tienen sólo
11 categorías cada una (mucho menos que las 716/748/789 categorías que tenían los códigos
ICD-9 originales antes de agruparse) — se incluyen igual en esta comparación porque, aun con
cardinalidad moderada, el target encoding sigue siendo una alternativa razonable frente al
one-hot, evaluando cuántas columnas genera cada método:

- **Frequency encoding:** reemplaza cada categoría por su frecuencia relativa en train. Muy
  simple y de una sola columna, pero **no usa el target** -sólo aporta señal si la frecuencia de
  un diagnóstico está en sí misma correlacionada con el reingreso (no necesariamente cierto).
- **Binary encoding:** codifica cada categoría como un número binario (una columna por dígito),
  reduciendo mucho la dimensionalidad frente a one-hot, pero sin relación directa con el target.

```python
# --- Frequency Encoding (a modo comparativo) ---
frecuencias_train = {}
for c in cols_alta_cardinalidad:
    frecuencias_train[c] = X_train[c].value_counts(normalize=True)

freq_encoded_ejemplo = X_train[["diag_1"]].copy()
freq_encoded_ejemplo["diag_1_freq"] = X_train["diag_1"].map(frecuencias_train["diag_1"])
freq_encoded_ejemplo.head()
```

```python
# --- Binary Encoding (a modo comparativo) ---
binary_encoder = BinaryEncoder(cols=["diag_1"])
diag_1_binary = binary_encoder.fit_transform(X_train[["diag_1"]])
print("Columnas generadas por Binary Encoding para 'diag_1':", diag_1_binary.shape[1])
diag_1_binary.head()
```

```python
# --- Comparación de dimensionalidad generada por cada técnica ---
comparacion_encoding = pd.DataFrame({
    "variable": cols_alta_cardinalidad,
    "n_categorias": [X_train[c].nunique() for c in cols_alta_cardinalidad],
    "cols_one_hot": [X_train[c].nunique() for c in cols_alta_cardinalidad],
    "cols_binary": [int(np.ceil(np.log2(X_train[c].nunique() + 1))) for c in cols_alta_cardinalidad],
    "cols_frequency": [1] * len(cols_alta_cardinalidad),
    "cols_target_enc": [1] * len(cols_alta_cardinalidad),
})
comparacion_encoding
```

**Lectura y decisión final:** para `diag_1` (11 categorías tras la agrupación por capítulo
ICD-9), Binary Encoding genera sólo 4 columnas en train -según se ve arriba-, y one-hot generaría
apenas 11: la diferencia de dimensionalidad entre métodos ya no es drástica para estas variables
en particular (sí lo sería para `medical_specialty`, con 72 categorías). Pero esas columnas
binarias son códigos sin significado clínico directo -dificulta la interpretabilidad- y, al no
usar el target, no concentran directamente la señal predictiva. Frequency encoding es aún más
compacto (1 columna), pero mezcla en un solo número diagnósticos distintos que simplemente
tengan una frecuencia similar. **Se mantiene target encoding** como la estrategia elegida para
estas variables: usa una sola columna por variable (igual que frequency encoding) pero, a
diferencia de esta, **incorpora directamente la relación con el target** -la tasa histórica de
reingreso de cada grupo diagnóstico-, que es la señal más relevante para el problema de
clasificación planteado, más allá de que la dimensionalidad en sí ya no sea un problema tan
grande para `diag_1/2/3` después de la agrupación.

```python
# --- One-hot encoding para variables nominales de baja cardinalidad ---
cols_baja_cardinalidad = ["gender", "A1Cresult", "max_glu_serum", "diabetesMed", "change",
                            "race", "payer_code", "admission_type_id"] + cols_familias_nuevas
cols_baja_cardinalidad = [c for c in cols_baja_cardinalidad if X_train[c].nunique() <= 15]

ohe = OneHotEncoder(handle_unknown="ignore", sparse_output=False)
ohe.fit(X_train[cols_baja_cardinalidad])

ohe_train = pd.DataFrame(ohe.transform(X_train[cols_baja_cardinalidad]),
                            columns=ohe.get_feature_names_out(cols_baja_cardinalidad), index=X_train.index)
ohe_test = pd.DataFrame(ohe.transform(X_test[cols_baja_cardinalidad]),
                           columns=ohe.get_feature_names_out(cols_baja_cardinalidad), index=X_test.index)

print("Columnas generadas por one-hot:", ohe_train.shape[1])
```

### 5.3 Ensamblado del dataset final de features numéricas

```python
cols_numericas_modelo = cols_numericas + ["age_ord", "patient_nbr_ordinal_ord"] + [c + "_te" for c in cols_alta_cardinalidad]

X_train_final = pd.concat([X_train[cols_numericas_modelo].reset_index(drop=True),
                             ohe_train.reset_index(drop=True)], axis=1)
X_test_final = pd.concat([X_test[cols_numericas_modelo].reset_index(drop=True),
                            ohe_test.reset_index(drop=True)], axis=1)

y_train_final = y_train.reset_index(drop=True)
y_test_final = y_test.reset_index(drop=True)

print("X_train_final:", X_train_final.shape, " X_test_final:", X_test_final.shape)
```

Se aplica `StandardScaler` (media 0, desvío 1), ajustado **sólo con train** y aplicado a
ambos conjuntos, ya que varios de los modelos que podrían usarse después (regresión logística,

```python
scaler = StandardScaler()
X_train_scaled = pd.DataFrame(scaler.fit_transform(X_train_final), columns=X_train_final.columns)
X_test_scaled = pd.DataFrame(scaler.transform(X_test_final), columns=X_test_final.columns)

X_train_scaled.describe().T[["mean", "std"]].round(2).head(10)
```

El desbalance es **moderado** (~89% / 11%), no extremo, pero suficiente para que un

### 6.2 Mecanismos de balanceo propuestos

Se evalúan tres alternativas, todas aplicadas **únicamente sobre el set de entrenamiento**
(nunca sobre test, para no distorsionar la evaluación):

1. **Class weighting** (`class_weight="balanced"` en el modelo): no modifica los datos, sólo
   penaliza más los errores sobre la clase minoritaria. Es la opción más simple y la que menos
   riesgo de overfitting introduce.
2. **Undersampling aleatorio** de la clase mayoritaria: reduce el dataset, rápido, pero
   descarta información potencialmente útil (no ideal con ~100k filas donde ya la clase
   minoritaria tiene ~8.0k casos, cantidad razonable).
3. **Oversampling con SMOTE** (Synthetic Minority Oversampling Technique): genera ejemplos
   sintéticos de la clase minoritaria interpolando entre vecinos cercanos, en vez de simplemente
   duplicar filas.

**Elección:** se aplica **SMOTE**, porque conserva toda la información de la clase mayoritaria
(a diferencia del undersampling) y evita el sobreajuste que produce duplicar exactamente los
mismos casos minoritarios (a diferencia de un oversampling naive). Se aplica **después** del
split y del encoding/escalado, y sólo sobre train.

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=RANDOM_STATE)
X_train_bal, y_train_bal = smote.fit_resample(X_train_scaled, y_train_final)

print("Antes de SMOTE:", y_train_final.value_counts().to_dict())
print("Después de SMOTE:", y_train_bal.value_counts().to_dict())
```

## 7. Reducción de dimensionalidad

### 7.1 Selección de features (relación entre variables)

Antes de reducir dimensiones con PCA, conviene revisar si hay variables redundantes. Se mira
la correlación entre las features numéricas ya construidas (incluyendo las de target encoding)
para detectar pares muy correlacionados (|r| > 0.85), candidatos a eliminar por redundancia.

```python
corr_features = X_train_scaled.corr().abs()
alta_corr = corr_features.where(np.triu(np.ones(corr_features.shape), k=1).astype(bool))
pares_redundantes = [(col, idx, alta_corr.loc[idx, col])
                       for col in alta_corr.columns for idx in alta_corr.index
                       if pd.notna(alta_corr.loc[idx, col]) and alta_corr.loc[idx, col] > 0.85]

print(f"Pares de features con correlación > 0.85: {len(pares_redundantes)}")
for a, b, v in pares_redundantes[:15]:
    print(f"  {a} <-> {b}: {v:.2f}")
```

No se detecta redundancia fuerte generalizada entre las features numéricas construidas: de
las ~67 features del dataset final, sólo aparecen **9 pares** con correlación > 0.85, y se
explican por tres motivos distintos:

- **Complementos binarios exactos** (r = 1.00, por construcción algebraica): `gender_Male`/
  `gender_Female`, `diabetesMed_Yes`/`diabetesMed_No`, `change_No`/`change_Ch` — son dummies
  mutuamente excluyentes de una variable binaria one-hot, no "redundancia" real sino la
  codificación misma.
- **Casi-complementos de variables de familia farmacológica** (r ≈ 0.92-0.97): los pares
  `_Steady`/`_No` de las 5 familias con al menos dos drogas (`familia_biguanidas`,
  `familia_sulfonilureas`, `familia_meglitinidas`, `familia_glitazonas`,
  `familia_alfa_glucosidasa`). Estas variables tienen 4 categorías posibles (`No`, `Steady`,
  `Up`, `Down`), no 2, así que no son complementos binarios forzados como los anteriores — la
  correlación tan alta se debe a que, en la práctica, casi todos los pacientes caen en `No` o
  `Steady` (`Up`/`Down` son minoritarios), lo que hace que ambas dummies se muevan de forma casi
  opuesta sin ser matemáticamente obligatorias.
- **Correlación genuina no forzada por el encoding**: `race_Caucasian` / `race_AfricanAmerican`
  (r=0.88). `race` tiene 5 categorías, así que tampoco hay complemento binario forzado; la
  correlación alta se explica porque esas dos categorías concentran ~94% de los casos (74.8% y
  18.9% respectivamente) — es una redundancia real, aunque moderada, entre esas dos dummies
  puntuales.

En ningún caso se trata de una relación inesperada o un error: son consecuencias directas de
cómo estas variables se distribuyen y de cómo se codificaron. Esto sugiere que la mayor parte
de las variables aporta información relativamente independiente, y que **PCA tiene sentido más
como técnica de compresión/visualización que como eliminación de duplicados evidentes**.

### 7.2 PCA (extracción de features)

```python
pca_full = PCA(random_state=RANDOM_STATE)
pca_full.fit(X_train_bal)

var_acumulada = np.cumsum(pca_full.explained_variance_ratio_)
n_comp_90 = np.argmax(var_acumulada >= 0.90) + 1
n_comp_95 = np.argmax(var_acumulada >= 0.95) + 1

plt.figure(figsize=(10, 5))
plt.plot(range(1, len(var_acumulada) + 1), var_acumulada, marker=".")
plt.axhline(0.90, color="orange", ls="--", label="90% varianza")
plt.axhline(0.95, color="red", ls="--", label="95% varianza")
plt.xlabel("Número de componentes principales")
plt.ylabel("Varianza acumulada explicada")
plt.title("Curva de varianza explicada por PCA")
plt.legend()
plt.show()

print(f"Componentes necesarios para explicar 90% de la varianza: {n_comp_90}")
print(f"Componentes necesarios para explicar 95% de la varianza: {n_comp_95}")
```

```python
pca_final = PCA(n_components=n_comp_90, random_state=RANDOM_STATE)
X_train_pca = pca_final.fit_transform(X_train_bal)
X_test_pca = pca_final.transform(X_test_scaled)

print(f"Dimensión original: {X_train_bal.shape[1]} features")
print(f"Dimensión tras PCA: {X_train_pca.shape[1]} componentes ({n_comp_90} de {X_train_bal.shape[1]})")
```

#### Varianza explicada por cada componente individual

Antes de graficar la proyección en 2D, se muestra qué porcentaje de la varianza total
aporta cada componente principal por separado (no acumulada). Es esperable que los
primeros componentes concentren la mayor parte y que el aporte caiga rápido después,
ya que **PCA** ordena los componentes de mayor a menor varianza explicada por construcción.

```python
varianza_individual = pca_final.explained_variance_ratio_ * 100

tabla_varianza = pd.DataFrame({
    "componente": [f"PC{i+1}" for i in range(len(varianza_individual))],
    "% varianza explicada": varianza_individual.round(2),
    "% varianza acumulada": np.cumsum(varianza_individual).round(2),
})

print(f"Las primeras 10 de {len(varianza_individual)} componentes (de {n_comp_90} en total, 90% de varianza):")
display(tabla_varianza.head(10))

plt.figure(figsize=(10, 5))
plt.bar(tabla_varianza["componente"][:15], tabla_varianza["% varianza explicada"][:15], color="steelblue")
plt.ylabel("% de varianza explicada")
plt.xlabel("Componente principal")
plt.title("Varianza explicada por cada componente (primeras 15)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### 7.3 Ventajas y desventajas de la reducción con PCA

**Ventajas observadas:**
- Reduce la dimensionalidad de forma moderada (de las features tras el one-hot encoding a
  un número algo menor de componentes manteniendo 90% de la varianza, ver salida de la celda
  anterior — la reducción no es drástica porque, como se vio en 7.1, la mayoría de las
  variables ya aporta información relativamente independiente), lo que igual acelera el
  entrenamiento y mitiga el efecto de la "maldición de la dimensionalidad" propia del
  one-hot encoding.
- Al ser combinaciones lineales ortogonales, elimina la multicolinealidad remanente entre
  features.

**Desventajas / trade-offs:**
- **Pérdida de interpretabilidad clínica**: cada componente principal es una combinación lineal
  de decenas de variables originales, por lo que ya no es posible decir "el reingreso aumenta
  por la variable X" — algo valioso en un dominio médico donde la explicabilidad importa tanto
  como la performance.
- La proyección 2D muestra que las clases **no quedan claramente separadas** en las primeras
  componentes, lo cual es esperable: el reingreso hospitalario depende de una combinación de
  múltiples factores débiles más que de 1-2 ejes dominantes, así que probablemente un modelo
  no lineal (Random Forest, Gradient Boosting) capture mejor esas interacciones que una
  proyección lineal.
- Al estar ajustado sobre datos balanceados con SMOTE, el PCA puede estar levemente sesgado
  hacia la estructura de los ejemplos sintéticos generados; se optó igualmente por este orden
  (split → encoding/escalado → SMOTE → PCA) para que el propio PCA "vea" un problema
  balanceado, coherente con el modelo que se entrenaría después.

## 8. Conclusiones e insights

**Perfil de paciente con mayor riesgo de reingreso temprano**, según la evidencia estadística
recolectada a lo largo del análisis:

- **Tipo de alta como principal asociación categórica**: `discharge_disposition_id` fue,
  de lejos, la variable categórica con mayor asociación al target (V de Cramér = 0.133, sección
  3.3), muy por encima del resto. `diag_1`, `diag_2`, `diag_3` (ya agrupados por capítulo ICD-9)
  y la edad no se destacan especialmente frente a otras variables — quedan en un grupo intermedio
  de asociación moderada-baja, junto con `admission_type_id`, `payer_code`, `insulin` y `change`.
  De hecho, la tasa de reingreso `<30` **no crece de forma sostenida** con la edad, sino que pica
  temprano (~14% en el grupo 20-30 años), baja hacia la mediana edad y vuelve a subir
  gradualmente después de los 60 (sección 3.3).
- **Historial reciente de internaciones/emergencias**: `number_inpatient` y
  `number_emergency`, aunque muy asimétricas, concentran a los pacientes de mayor riesgo — son
  justamente las variables cuyos valores extremos se decidió **winsorizar** (recortar al
  percentil 99) en la sección 2.3 en lugar de eliminar directamente esas filas, porque esos
  valores altos son la señal más relevante del problema.
- **Internaciones más largas y con más medicación/procedimientos**: `time_in_hospital`,
  `num_medications` y `num_lab_procedures` están moderadamente asociadas entre sí y reflejan
  cuadros clínicos más complejos, coherentes con mayor probabilidad de reingreso.
- **Tipo de alta (`discharge_disposition_id`)**: pacientes dados de alta hacia otras
  instituciones (vs. alta a domicilio) muestran mayor tasa de reingreso, algo esperable ya que
  refleja peor estado general al momento del alta.
- Variables administrativas como `gender` o `payer_code` mostraron muy poca asociación con el
  reingreso, y no parecen buenos predictores por sí solas.

**Lecciones del proceso de preparación de datos:**
1. Gran parte del trabajo de limpieza consistió en **distinguir faltantes estructurales
   (MNAR)** — como `weight` o `A1Cresult`, que no se miden salvo indicación médica — **de
   faltantes por proceso administrativo (MAR)** como `medical_specialty`, y tratarlos de forma
   distinta (columna descartada vs. categoría "Faltante"/"No medido" explícita).
2. En datos clínicos, un "outlier" estadístico (según IQR) frecuentemente **es** la observación
   clínicamente más importante (pacientes de alto riesgo); eliminarlo automáticamente hubiese
   sesgado el problema en la dirección opuesta a la que se busca predecir.
3. Para variables categóricas con cardinalidad media-alta (`medical_specialty` con 72
   categorías; `diag_1/2/3`, que quedaron en 11 categorías cada una tras agruparse por capítulo
   ICD-9, pero conservando señal relevante por grupo diagnóstico) el target encoding calculado
   exclusivamente en train permitió capturar directamente la tasa de reingreso de cada categoría
   sin filtrar información del set de prueba.
4. El desbalance de clases (~89/11) es moderado pero suficiente para requerir un tratamiento
   explícito (SMOTE) antes de entrenar cualquier modelo, para no obtener un clasificador que
   ignore a los pacientes de riesgo.
5. PCA resultó útil para comprimir la dimensionalidad generada por el one-hot encoding, aunque
   a costa de interpretabilidad clínica — un trade-off a evaluar según si el objetivo final
   prioriza performance predictiva pura o explicabilidad para el equipo médico.
